This takes the variable 'status' and maps particular strings to the values (see below) as a dictionary.

```python
df.head()
```

```python
status_values = {
    1: 'ok', 
    2: 'default', 
    0: 'unk'
}
df.status = df.status.map(status_values)

home_values = {
    1: 'rent', 
    2: 'owner', 
    3: 'private',
    4: 'ignore',
    5: 'parents',
    6: 'other',
    0: 'unk'
}
df.home = df.home.map(home_values)

marital_values = {
    1: 'single', 
    2: 'married', 
    3: 'widow',
    4: 'separated',
    5: 'divorced',
    0: 'unk'
}
df.marital = df.marital.map(marital_values)

records_values = {
    1: 'no', 
    2: 'yes', 
    0: 'unk'
}
df.records = df.records.map(records_values)

job_values = {
    1: 'fixed', 
    2: 'partime', 
    3: 'freelance',
    4: 'others',
    0: 'unk'
}
df.job = df.job.map(job_values)
```

```python
for c in ['income', 'assets', 'debt']:
    df[c] = df[c].replace(to_replace=99999999, value=np.nan)
```

We're only interested in records that are known. Therefore, we filter out 'unk'.

6.3 Decision Trees

* How a decision tree looks
* Training a decision tree
* Overfitting
* Controlling the size of a tree

We use the dataset we prepared previously to see if customers will default or not.

```python
def assess_risk(client):
    if client['records'] == 'yes':
        if client['job'] == 'parttime':
            return 'default'
        else:
            return 'ok'
    else:
        if client['assets'] > 6000:
            return 'ok'
        else:
            return 'default'
```

```python
df_train.iloc[0].to_dict()
```

```python
xi = df_train.iloc[0].to_dict()
```

```python
assess_risk(xi)
```

```python
xi
```

They have no 'records' but have 10000 in assets, so risk assessment is 'ok'.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.feature_extraction import DictVectorizer
```

```python
train_dicts = df_train.to_dict(orient='records')
```

```python
train_dicts = df_train.fillna(0).to_dict(orient='records')
```

```python
train_dicts[:5]
```

```python
dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(train_dicts)
```

```python
X_train
```

This is our feature matrix.

```python
dv.get_feature_names_out()
```

We have encoding for categorical features while the numerical features are left intact. We can now train the decision tree.

```python
dt = DecisionTreeClassifier()
dt.fit(X_train, y_train)
```

Now we have our tree, so let's test it, using the validation dataset. We create X_val (validation using dictionary vectoriser) and then transform.

```python
val_dicts = df_val.fillna(0).to_dict(orient='records')
X_val = dv.transform(val_dicts)
```

```python
y_pred = dt.predict_proba(X_val)
```

y_pred is our decision tree. Next we calculate AUC score...

```python
from sklearn.metrics import roc_auc_score
```

```python
y_pred
```

```python
y_pred = dt.predict_proba(X_val)[:,1]
```

We need take the second column i.e. the probability of the customer belonging to the second class.

```python
roc_auc_score(y_val, y_pred)
```

AUC score = 65 - not impressive.

Let's have a look at the training dataset.

```python
 y_pred = dt.predict_proba(X_train)[:,1]
roc_auc_score(y_train, y_pred)
```

Training datset = 1. In other words the model is overfitting (it memorises the data but fails to generalise on a new data point and outputs something completely incorrect).

This can happen when the decision tree becomes quite deep and the model learns any possible combination for any customer and then at the end the decision can be just for one or two clients, which is then memorised into the model.

```python
dt = DecisionTreeClassifier(max_depth=3)
dt.fit(X_train, y_train)
```

```python
y_pred = dt.predict_proba(X_train)[:,1]
auc = roc_auc_score(y_train, y_pred)
print('train:', auc)

y_pred = dt.predict_proba(X_val)[:,1]
roc_auc_score(y_val, y_pred)
print('val:', auc)
```

After restricting the depth to 'three', we can see the performance of the model on validation is signicantly better: 77 rather than 65. When restricted to a depth of \one', then it becomes a 'decision stump' and the result is not significantly worse than the original result (62 vs 63).

```python
dt = DecisionTreeClassifier(max_depth=1)
dt.fit(X_train, y_train)
```

```python
y_pred = dt.predict_proba(X_train)[:,1]
auc = roc_auc_score(y_train, y_pred)
print('train:', auc)

y_pred = dt.predict_proba(X_val)[:,1]
roc_auc_score(y_val, y_pred)
print('val:', auc)
```

We can have a look inside the tree and see what rules it learned. 

```python
from sklearn.tree import export_text
```

```python
print(export_text(dt, feature_names=dv.get_feature_names_out()))
```

It shows that if the feature 25 is less than 0.5, it predicts default, or greater than 0.5 then it's ok.

```python
dt = DecisionTreeClassifier(max_depth=3)
dt.fit(X_train, y_train)
y_pred = dt.predict_proba(X_train)[:,1]
auc = roc_auc_score(y_train, y_pred)
print('train:', auc)

y_pred = dt.predict_proba(X_val)[:,1]
roc_auc_score(y_val, y_pred)
print('val:', auc)
print(export_text(dt, feature_names=dv.get_feature_names_out()))
```
